1. Create a NumPy array of size 2 × 2

```python
import numpy as np

arr = np.array([[1, 2],
                [3, 4]])

print(arr)
print("Shape:", arr.shape)
```

2. Create a random NumPy array of size 3 × 4

```python
arr_random = np.random.rand(3, 4)

print(arr_random)
print("Shape:", arr_random.shape)
```

3. Create a NumPy array of ones and a NumPy array of zeroes

```python
arr_ones = np.ones((3, 3))
arr_zeroes = np.zeros((3, 3))

print("Array of ones:")
print(arr_ones)

print("Array of zeroes:")
print(arr_zeroes)
```

4. Create a random NumPy array of size 6 × 3 and transpose it

```python
arr = np.random.rand(6, 3)

print("Original array:")
print(arr)
print("Original shape:", arr.shape)

arr_transposed = arr.T

print("Transposed array:")
print(arr_transposed)
print("Transposed shape:", arr_transposed.shape)
```

5. Create two random NumPy arrays of size n × p and p × m, then compute the dot product

```python
n = 2
p = 3
m = 4

A = np.random.rand(n, p)
B = np.random.rand(p, m)

print("Array A:")
print(A)
print("Shape of A:", A.shape)

print("Array B:")
print(B)
print("Shape of B:", B.shape)

result = np.dot(A, B)

print("Dot product:")
print(result)
print("Shape of result:", result.shape)
```
